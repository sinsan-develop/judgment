저장소 루트에서 실행하며 원문을 출력하거나 추출하지 않는다. 표준 라이브러리와 이미 제공된 openpyxl 환경을 사용한다. 학습·네트워크 호출은 없다.

```python
import importlib.util
import json
from pathlib import Path
spec = importlib.util.spec_from_file_location("review", Path("docs/aihub116_review.py"))
review = importlib.util.module_from_spec(spec)
spec.loader.exec_module(review)
rows, profiles = review.ingest(Path("docs/한국어대화_new_260226.zip"))
fixed_map = json.loads(Path(".work/data_review/group_map.private.json").read_text(encoding="utf-8"))
summary, private_map = review.analyze(rows, profiles, full_mapping=fixed_map)
{k: summary[k] for k in ("total_rows", "unique_input_examples", "distinct_utterances", "intent_count", "group_count", "split", "pilot")}
```
